```r
library(dplyr)
library(ggplot2)
library(gridExtra)
options(repr.plot.width = 20, repr.plot.height = 6)
```

```r
f <- function(x) {-sin(x) - exp(x / 100) + 10}
f1 <- function(x) {-sin(x) - exp(x / 100) + 10.3 + 0.03*(x-3)^2}
f2 <- function(x) {-sin(x) - exp(x / 100) + 10.3 + 0.01*(x-3)^2}
f3 <- function(x) {-sin(x) - exp(x / 100) + 10.3 + 0.05*(x-3)^2}
f4 <- function(x) {-sin(x) - exp(x / 100) + 10.3}
f5 <- function(x) {-sin(x) - exp(x / 100) + 10.3 - 0.03*(x-3)^2}
f6 <- function(x) {-sin(x) - exp(x / 100) + 10.3 + 0.03*(x-7)^2}
f7 <- function(x) {-sin(x) - exp(x / 100) + 10.3 + 0.07*(x-3)^2}
f8 <- function(x) {-sin(0.5*x+11) - exp((0.5*x+11) / 100) + 10}
f9 <- function(x) {-sin(x+2) - exp((x+2) / 100) + 10}
f10 <- function(x) {10.3 + 0.01*(x-7)^2}
f11 <- function(x) {-sin(x) - exp(x / 100) + 10.3 + 0.03*(x-3)^2}
f12 <- function(x) {-sin(x) - exp(x / 100) + 10.3 + 0.01*(x-3)^2}
```

```r
plot <- function(exp,m,i,func) {
    df <- read.table(paste0("./exp", exp, "/results/data_macrorep_", m, "_iteration_", i, ".txt"), header = TRUE)

    dp_hf <- df %>% filter(hf==1) %>% mutate(yH=f(x),yL=func(x))
    dp_lf <- df %>% filter(lf==1) %>% mutate(yH=f(x),yL=func(x))
    df <- df %>% mutate(yH=f(x), yL=func(x))

    sf <- max(df$yH, na.rm = TRUE) / max(abs(df$yH - df$yL), na.rm = TRUE)
    
    p1 <- ggplot(df, aes(x,yH)) +
      geom_line() +
      geom_line(aes(x,yL),colour="blue") +
      geom_line(aes(x,(yH-yL)*sf),colour="orange") + 
      geom_point(
        data = dp_lf,
        aes(x,yL),
        shape = 4,        # cross
        colour = "orange",
        size = 3,
        stroke = 1
      )   +
      geom_point(
        data = dp_hf,
        aes(x,yH),
        shape = 4,        # cross
        colour = "green",
        size = 3,
        stroke = 1
      )  +
      labs(title = paste0("Macrorep: ", m, ", Iteration: ", i, "\n \n High Fidelity Surface"),
      x = "x",
      y = "y") +
      theme_minimal()

    p2 <- ggplot(df, aes(x = x, y = cond_mean)) +
        geom_ribbon(
        aes(
          ymin = cond_mean - 2 * sqrt(cond_var),
          ymax = cond_mean + 2 * sqrt(cond_var)
        ),
        alpha = 0.15
      ) +
      geom_line() +
      geom_point(
        data = dp_lf,
        aes(x,y),
        shape = 4,        # cross
        colour = "orange",
        size = 3,
        stroke = 1
      )   +
      geom_point(
        data = dp_hf,
        aes(x,y),
        shape = 4,        # cross
        colour = "green",
        size = 3,
        stroke = 1
      )  +
      labs(title = "Conditional mean",
      x = "x",
      y = "Conditional mean") +
      theme_minimal()

    p3 <- ggplot(df, aes(x = x, y = cond_mean_err)) +
        geom_ribbon(
        aes(
          ymin = cond_mean_err - 2 * sqrt(cond_var_err),
          ymax = cond_mean_err + 2 * sqrt(cond_var_err)
        ),
        alpha = 0.15
      ) +
      geom_point(
        data = dp_lf,
        aes(x,y),
        shape = 4,        # cross
        colour = "orange",
        size = 3,
        stroke = 1
      )   +
      geom_point(
        data = dp_hf,
        aes(x,y),
        shape = 4,        # cross
        colour = "green",
        size = 3,
        stroke = 1
      )  +
      labs(title = "Conditional mean error",
      x = "x",
      y = "Conditional mean error") +
      theme_minimal()    

    p4 <- ggplot(df, aes(x = x, y = cei)) +
      geom_line() +
      geom_point(
        data = dp_lf,
        aes(x,y),
        shape = 4,        # cross
        colour = "orange",
        size = 3,
        stroke = 1
      )   +
      geom_point(
        data = dp_hf,
        aes(x,y),
        shape = 4,        # cross
        colour = "green",
        size = 3,
        stroke = 1
      )  +
      labs(title = "CEI",
      x = "x",
      y = "CEI") + 
      theme_minimal()

    grid.arrange(p1,p2,p3,p4,ncol=4)   
}
```

```r
for (i in 1:20) {
  plot("051",0,i, f1)
}
```

```r
for (i in 1:20) {
  plot("070",2,i, f2)
}
```
