```r
library(dplyr)
library(ggplot2)
library(gridExtra)
options(repr.plot.width = 20, repr.plot.height = 6)
```

```r
df_hf <- read.table("./data_hf_1000_rep.txt", header = TRUE)
```

```r
plot <- function(i) {
    df <- read.table(paste0("./results/data_iteration_", i, ".txt"), header = TRUE)
    df <- df %>%
        left_join(df_hf,
                  by = c('x'='solx','y'='soly'))  

    p1 <- ggplot(df, aes(x = y, y = x, fill = output)) +
      geom_raster() +
      labs(title = "High Fidelity Surface",
      x = "S - s",
      y = "s") +
      geom_tile(
          data = dplyr::filter(df, output == min(output, na.rm = TRUE)),
          color = "white",
          linewidth = 0.5,
          fill = NA) + 
      theme_minimal()

    p2 <- ggplot(df, aes(x = y, y = x, fill = cond_mean)) +
      geom_raster() +
      labs(title = "Conditional mean",
      x = "S - s",
      y = "s") +
      geom_tile(
          data = dplyr::filter(df, lf==1),
          color = "orange",
          linewidth = 0.5,
          fill = NA) +
      geom_tile(
          data = dplyr::filter(df, hf==1),
          color = "green",
          linewidth = 0.3,
          fill = NA) +
      theme_minimal()

    p3 <- ggplot(df, aes(x = y, y = x, fill = cei)) +
      geom_raster() +
      labs(title = "CEI",
      x = "S - s",
      y = "s") +
      geom_tile(
          data = dplyr::filter(df, x_star == 1),
          color = "white",
          linewidth = 0.5,
          fill = NA) + 
      geom_tile(
          data = dplyr::filter(df, cei == max(cei, na.rm = TRUE) & lf_next == 1),
          color = "orange",
          linewidth = 0.5,
          fill = NA) +
      geom_tile(
          data = dplyr::filter(df, cei == max(cei, na.rm = TRUE) & hf_next == 1),
          color = "green",
          linewidth = 0.5,
          fill = NA) +    
      theme_minimal()

    grid.arrange(p1,p2,p3,ncol=3)   
}
```

```r
for (i in 1:100) {
  plot(i)
}
```
