```r
library(dplyr)
library(ggplot2)
library(gridExtra)
options(repr.plot.width = 20, repr.plot.height = 6)
```

```r
df_hf <- read.table("./data_hf_1000_rep_incl_std.txt", header = FALSE)
colnames(df_hf) <- c("x","y","mean","std")
```

```r
ggplot(df_hf, aes(x = y, y = x, fill = mean)) +
  geom_raster() +
  labs(title = "High Fidelity Surface",
  x = "S - s",
  y = "s") +
  theme_minimal()
```

```r
ggplot(df_hf, aes(x = y, y = x, fill = std)) +
  geom_raster() +
  labs(title = "High Fidelity Surface",
  x = "S - s",
  y = "s") +
  theme_minimal()
```
